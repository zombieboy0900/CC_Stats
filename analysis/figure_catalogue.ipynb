# Figure catalogue: every figure not in the story

`story.ipynb` holds the figures picked for the presentation. This notebook holds **every other figure** the analysis folders draw, grouped by the question they belong to, each with what it shows and where it comes from. It's for backup slides, questions, and choosing a different figure for a slide. Each outcome figure appears twice: Engel, then ILAE. Each section ends with the same figures by seizure code (`codes/` notebooks).

The figures are the PNGs the folder notebooks save (`results/<version>/<folder>/figures/`, via `cc_stats.plots.save_figure`), so they are exactly what those notebooks draw. Run `pixi run run-analysis` first; `run_all.py` runs this notebook last. A figure marked *as in the story* is the folder's own version of a story figure, kept here so every saved figure is listed.

```python
from IPython.display import Image, display

from cc_stats.paths import DATASET, results_dir

print(f'dataset {DATASET}')


# A figure a folder notebook saved (see cc_stats.plots.save_figure).
def show(folder, name):
    path = results_dir(folder) / 'figures' / f'{name}.png'
    if not path.exists():
        raise FileNotFoundError(f'{path} not found - run `pixi run run-analysis` first')
    display(Image(filename=path, width=700))
```
